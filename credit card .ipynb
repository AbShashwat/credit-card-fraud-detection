```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
card_data=pd.read_csv('creditcard.csv.csv')
```

```python
card_data.head()
```

```python
card_data.tail()
```

```python
card_data.info()
```

```python
card_data.isnull().sum()
```

```python
card_data['V25']=card_data['V25'].fillna(card_data['V19'].mean())
card_data['V26']=card_data['V26'].fillna(card_data['V26'].mean())
card_data['V26']=card_data['V26'].fillna(card_data['V26'].mean())
card_data['V27']=card_data['V27'].fillna(card_data['V27'].mean())
card_data['V28']=card_data['V28'].fillna(card_data['V28'].mean())
card_data['Amount']=card_data['Amount'].fillna(card_data['Amount'].mean())
card_data['Class']=card_data['Class'].fillna(card_data['Class'].mean())
```

```python
card_data.isnull().sum()
```

```python
card_data['Class'].value_counts()
```

```python
 legit=card_data[card_data.Class==0]
 fraud=card_data[card_data.Class==1]
```

```python
print(legit.shape)
print(fraud.shape)
```

```python
legit.Amount.describe()
```

```python
fraud.Amount.describe()
```

```python
 card_data.groupby('Class').mean()
```

```python
legit_sample=legit.sample(n=492)
```

```python
new_dataset=pd.concat([legit_sample,fraud],axis=0)
```

```python
new_dataset.head()
```

```python
new_dataset['Class'].value_counts()
```

```python
X=new_dataset.drop(columns='Class', axis=1)
Y=new_dataset['Class']
```

```python
print(X)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.20,stratify=Y, random_state=2)
```

```python
print(X_train, X_test, Y_train, Y_test)
```

```python
model = LogisticRegression()
model.fit(X_train, Y_train)
```

```python
X_train_prediction=model.predict(X_train)
training_data_accuracy=accuracy_score(X_train_prediction,Y_train)
```

```python
print('training_data_accuracy',training_data_accuracy)
```

```python
X_test_prediction=model.predict(X_test)
test_data_accuracy=accuracy_score(X_test_prediction,Y_test)
```

```python
print('Accuracy score on test data:',test_data_accuracy)
```
